## 4.2 Masking

Another technique for selecting relevant information (which will be often useful) is called **masking**. In this scenario, 
- we define a boolean expression (i.e. one which evaluates to `True` or `False`)
- apply it to a column
- select all rows for which the mask evaluates to `True`.

Below we show how this works for selecting data within a certain date range.

````python
# first we inspect the data type
df.year.dtype
````

````python
# we formulate a boolean expression
year = 1899
print('year > 1900 = ', year > 1900)
print('year < 1900 = ', year < 1900)
````

We can apply this expression to a column, which returns a series with the boolean value for each row index.
This series we use as a **mask** for selecting rows in the dataframe.

````python
df['year'] > 1900
````

In Python `False` is interpreted as zero and `True` as 1. To measure how many rows match a condition we can apply `sum` to a mask.

````python
# we have 1802468 row with names after 1900
sum(df['year'] > 1900)
````

````python
# saving subset in a new dataframe
# note that index does not start from zero!
mask = df.year > 1900
df_after_1900 = df[mask]
df_after_1900
````

````python
sum(df['year'] > 1900) == df_after_1900.shape[0]
````

## 4.3 Finding Information: Count and Select

#### Searching for love, love.

Pandas provides powerful functions to search through text and then select rows based on the number of hits. 

Below we show to:
- use the `count()` method to search for the string 'love' in the track titles
- use a mask to select all songs that mention 'love' at least twice

````python
# with applying count to a column we can count the number of occurences of a specific string
df_muse['track'].str.count('love')
````

````python
# we can also count the number of occurences of a specific string in a case-insensitive way
df_muse['track'].str.lower().str.count('love')
````

````python
# we cas save the result (search ignoring capitalization) in a new column
df_muse['hits'] = df_muse['track'].str.lower().str.count('love')
````

````python
# we can create a mask to filter the rows with more than one occurence of the word 'love'
mask = df_muse['hits'] > 1
mask
````

````python
# we can apply the mask to the dataframe
df_muse[mask]
````
