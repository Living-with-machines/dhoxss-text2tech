## Paths

The path is the location of an item (a file or a folder) in a directory structure.

In order to read a file from a Python script or notebook, we need to indicate the path of the file we want to read.

Paths can be provided as absolute paths or as a relative paths.

* **Absolute path:** The absolute path is the complete "address" of the file or folder in your computer, starting from the root. The absolute path will be different in each machine. For example, in my machine, the absolute path to the file `tale_beginning.txt` is:
  ```
  /Users/myself/Documents/LwM/dhoxss-text2tech/Sessions/data/tale_beginning.txt
  ```

* **Relative path:** The relative path is the location of a file or folder relative to the location of the python script or notebook that we are running. For example, from this notebook, the relative path to the file `tale_beginning.txt` is:
  ```
  data/tale_beginning.txt
  ```
  
  This means that, inside the same directory where I have this notebook, there is a folder named `data`, inside which there is a file named `tale_beginning.txt`. Note: If you needed to go one level back in the folder structure, you can do it with `../`.

To know more about paths and working locally, check out Melanie Walsh's [great introduction to understanding the command line](https://melaniewalsh.github.io/Intro-Cultural-Analytics/01-Command-Line/01-The-Command-Line.html).

Once you know the path to your file, you can open it using the following syntax:

```python
with open("path_to_file") as fr:
    text = fr.read()
```

Where:
* The first line opens the file (you should change `path_to_file` to the actual path to your file) and keeps it temporarily in variable `fr`.
* The second line (note that it's indented), fetches the actual text from variable `fr` and puts it in variable `text`.

For example, to read the content of `tale_beginning.txt`:

````python
# Read the content of "tale_beginning.txt":
with open("data/tale_beginning.txt") as fr:
    text = fr.read()
````
