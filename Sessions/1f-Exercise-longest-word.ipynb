# 👀 Suggested solutions

✏️ **Exercise 1: find the longest word with a loop**

````python
longest_word = ""                       # we haven't seen any word yet

for word in words:
    if len(word) > len(longest_word):   # is this word longer than the best one so far?
        longest_word = word             # if so, it becomes the new best one

print(longest_word)
print(len(longest_word))
````

✏️ **Exercise 2: turn it into a function**

````python
def find_longest_word(list_of_words):
    """ return the longest word in a list of words """
    longest_word = ""
    for word in list_of_words:
        if len(word) > len(longest_word):
            longest_word = word
    return longest_word


# first, a small test where we know the answer:
print(find_longest_word(["it", "was", "the", "best", "of", "times"]))

# and now the whole book:
print(find_longest_word(words))
````

❓ **Question:** the "word" we get is something like `letter-writing?—delicate—tender—truly`: a whole series of words that Austen (or her typesetter) joined with hyphens and dashes, with no spaces in between. Since `.split()` only cuts on whitespace, Python sees all of that as a single word.

This is not a bug in our function — our function is right, our *definition of a word* is too naive. Tomorrow we will see how to split text in better ways.

**Bonus 1: how long is it?**

````python
def find_longest_word_length(list_of_words):
    """ return the length of the longest word in a list of words """
    longest_word = find_longest_word(list_of_words)   # reusing the function we already wrote!
    return len(longest_word)


print(find_longest_word_length(["it", "was", "the", "best", "of", "times"]))
print(find_longest_word_length(words))
````

**Bonus 2: ties**

````python
maximum_length = find_longest_word_length(words)

counter = 0
for word in words:
    if len(word) == maximum_length:
        counter += 1
        print(word)

print("Number of words of maximum length:", counter)
````

**Bonus 3: the other end**

Starting from an empty string does not work, because no word can ever be *shorter* than a string of length 0, so the answer would always be `""`. A simple solution is to start from the first word of the list and compare everything against it.

````python
def find_shortest_word(list_of_words):
    """ return the shortest word in a list of words """
    shortest_word = None                  # nothing seen yet
    for word in list_of_words:
        if shortest_word == None:         # the very first word we look at
            shortest_word = word
        elif len(word) < len(shortest_word):
            shortest_word = word
    return shortest_word


print(find_shortest_word(["it", "was", "the", "best", "of", "times"]))
print(find_shortest_word(words))
````

**Bonus 4: empty input**

`find_longest_word([])` returns an empty string: the loop never runs, so the function returns the value we started with. That is a bit misleading — it looks as if the longest word of our text were `""`. Better to stop with a clear message:

````python
def find_longest_word(list_of_words):
    """ return the longest word in a list of words

    Args:
        list_of_words: a list of words (strings)

    Returns:
        The longest word in the list
    """
    assert type(list_of_words) is list, 'The input is not a list'
    assert len(list_of_words) > 0, 'The input list is empty'

    longest_word = ""
    for word in list_of_words:
        if len(word) > len(longest_word):
            longest_word = word
    return longest_word


print(find_longest_word(words))
print(find_longest_word([]))    # this one now stops with a clear message
````
