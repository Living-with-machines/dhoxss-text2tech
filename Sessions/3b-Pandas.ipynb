## Selecting rows: masks

To select *rows*, we write a condition. The condition produces a column of `True`/`False` or boolean values.

This is called a **mask** — and pandas keeps the rows evaluate to `True`.

```python
# Which rows are in English?
mask = df["language"] == "English"

print(mask.head())
```

You can perform operation on the **mask**, for example to count how many English language books we have in the dataframe.

`True` counts as 1, so summing counts the matches

```python

print("English books:", mask.sum()) 
```

Putting the mask in square brackets gives us those rows as a new DataFrame:

```python
english = df[df["language"] == "English"]

print(english.shape)
english.head(3)
```

Conditions can be combined with `&` (and) and `|` (or). Each condition needs its own parentheses:

```python
early = df[(df["date"] < 1800) & (df["language"] == "English")]

print("English books published before 1800:", len(early))
early[["title", "date", "place"]].head()
```

☝️ Note `df[["title", "date", "place"]]`: a *list* of column names inside the brackets selects several columns at once.

## Working with text in a column

Here is where the week comes back together. Every string method you learnt in notebook 2a — `.lower()`, `.split()`, `.count()`, `.startswith()` — exists in pandas as `.str.something`, and applies to **the whole column at once**.

```python
# How many characters in each title?
df["title"].str.len().head()
```

```python
# How many words in each title? (.split() then count the pieces)
df["title"].str.split().str.len().head()
```

We can store the result as a **new column**, simply by assigning to a name that does not exist yet:

```python
df["title_words"] = df["title"].str.split().str.len()

df[["title", "title_words"]].head()
```

`.str.contains()` searches inside every title at once. It returns a mask, so we can use it to select rows:

```python
has_history = df["title"].str.contains("history", case=False)

print("titles containing 'history':", has_history.sum())
df[has_history][["title", "genre"]].head()
```

### `.apply()`: your own functions, over a whole column

`.str` covers the common operations. When you need something it does not offer, you write a **function** — exactly as in notebook 1d — and hand it to `.apply()`, which runs it on every row of the column.

```python
def first_word(title):
    """ return the first word of a title, lowercased, without punctuation

    Args:
        title: the title of a book, as a string

    Returns:
        The first word, as a string
    """
    words = title.lower().split()
    return words[0].strip(".,;:'\"")


# Try it on a single title first — always do this before applying it to 1,750 of them:
print(first_word("The Canadian farmer. A missionary incident"))
```

Now apply it to the whole column:

```python
df["first_word"] = df["title"].apply(first_word)

df["first_word"].value_counts().head(10)
```

⚠️ `.apply()` is the general escape hatch: anything you can write as a function, you can run over a whole column. It is slower than the built-in `.str` methods, so prefer those when they exist — but `.apply()` is what makes pandas open-ended rather than a fixed menu.

## Grouping: asking questions of subsets

`.groupby()` splits the table into groups, computes something for each, and puts the answers back together. It is the single most useful thing pandas does — and the picture explains it better than the words:

![split-apply-combine](https://jakevdp.github.io/PythonDataScienceHandbook/figures/03.08-split-apply-combine.png)

*(Figure from Jake VanderPlas, [Python Data Science Handbook](https://jakevdp.github.io/PythonDataScienceHandbook/).)*

**Split** the rows into groups, **apply** a calculation to each group, **combine** the answers into one table.

For example: do fiction and non-fiction titles differ in length?

```python
df.groupby("genre")["title_words"].mean()
```

They do — and in a direction worth thinking about. Now: how is genre distributed across languages?

```python
# `.isin()` keeps rows whose value is in a list — handy when a column has a long tail:
main_languages = ["English", "Dutch", "Swedish", "Danish", "German"]

df[df["language"].isin(main_languages)].groupby("language")["genre"].value_counts()
```

Look at that table closely, because it matters more than it looks.

The Danish books are 164 non-fiction to **1** fiction. The Swedish ones are 188 non-fiction to **none at all**. The English part of the collection is a mix; the non-English part is almost entirely non-fiction.

So in this collection, *language* and *genre* are tangled together. Any conclusion we draw about genre might really be a conclusion about language. Keep this in your pocket — in the next notebook it comes back to bite us.

### Trends over time

To look at change over time we need decades rather than individual years. Integer division (`//`) rounds down, so `1857 // 10 * 10` gives `1850`:

```python
df["decade"] = (df["date"] // 10) * 10

books_per_decade = df.groupby("decade").size()
print(books_per_decade.tail(10))
```

```python
# pandas can plot directly:
books_per_decade.plot(kind="bar", figsize=(12, 4), title="Books in the collection, by decade")
```

⚠️ Read that chart carefully. It does **not** show that more books were written in the 1880s. It shows that more books *from* the 1880s are in **this collection** — a fact about the British Library's digitisation choices, not about the nineteenth century.

```python
def is_long_title(title):
    """ True if the title has more than 15 words """
    return len(title.split()) > 15


df["long_title"] = df["title"].apply(is_long_title)
print(df["long_title"].value_counts())
```

### ✏️ Exercise 6

```python
def averge_word_length(text):
    words = text.split()
    word_lengths = [len(w) for w in words]
    return sum(word_lengths) / len(word_lengths)
    
df['av_word_length'] =df['title'].apply(averge_word_length)
df[['title','av_word_length']].sort_values(by='av_word_length', ascending=False)
```
