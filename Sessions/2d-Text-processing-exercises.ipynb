## Exercise: How many chapters are there in the novel?

Suggested steps (but feel free to be innovative):
1. Create a variable that will store the counting, instantiate it with 0 (number, not string!).
2. Iterate over each line of the novel.
3. For each line:
4. Check if the line indicates a new chapter (tip: use regular expressions).
5. If the line indicates a new chapter, increment the counter variable by 1.
5. Finally, when you are done, print the counter variable.
6. Go to the actual book, and check if you got the answer right.

```python
# Type your code here:

```

👀 **Suggested solution:**

```python
import re

counter = 0
pattern = r'^CHAPTER [A-Z]+\.$'
for line in booklines:
    myMatch = re.match(pattern, line)
    if myMatch:
        counter += 1
print(counter)
```

## Exercise: Return all the characters in the novel of a certain title

Find all occurrences of the word 'Miss' (or 'Mr.' or 'Mrs." or 'Lady' or 'Lord") followed by any name, and return a counter (i.e. how many times they occur in the novel).

Suggestions:
* Use the full text, not separated in lines (i.e. the `book` variable).
* Design a regular expression and use the `re.findall()` method to find occurrences of the pattern in the text.
* Use `Counter` from `collections` to count how many times character names occur in the novel.

```python
# Type your code here:

```

👀 **Suggested solution:**

```python
from collections import Counter

pattern = r'\bMiss [A-Z][a-z]+\b'
# # Other examples:
# pattern = r'\bMr\. [A-Z][a-z]+\b'
# pattern = r'\bMrs\. [A-Z][a-z]+\b'
# pattern = r'\bLady [A-Z][a-z]+\b'
# pattern = r'\bLord [A-Z][a-z]+\b'

list_characters = re.findall(pattern, book)
count_characters = Counter(list_characters)
```

```python
for k in count_characters:
    print(k, count_characters[k])
```

👀 **Suggested solution using a function:**

```python
from collections import Counter

def find_characters_by_title(title):
    pattern = title + r'\.? [A-Z][a-z]+\b'
    list_characters = re.findall(pattern, book)
    count_characters = Counter(list_characters)
    return count_characters
```

```python
find_characters_by_title("Mrs")
```

```python
find_characters_by_title("Miss")
```
