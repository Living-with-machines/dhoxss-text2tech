# Running LLMs on your laptop

Large language models are huge models that require a large amount of computational resources not only for training but also for generating an output.

![](images/llama3-meme.png)

## So, how do people manage to run them on their laptops?

### Quantization

- Reduces model size by converting weights from float32 to lower precision (e.g., int8).
- Improves inference speed and reduces memory footprint.
- Trade-off: May slightly impact model accuracy.

### Model Pruning

- Eliminates redundant or less important neurons/weights.
- Results in a sparser model that requires less computation.
- Trade-off: Can lead to a decrease in model performance if overly aggressive.

### Distillation

- Creates a smaller, more efficient model (student) that mimics a larger model (teacher).
- Retains much of the original model’s performance with fewer parameters.
- Useful for deploying models on resource-constrained devices.

### Model Compression

- Apply methods such as weight sharing and low-rank factorization to compress the model.
- Helps in fitting larger models into limited laptop storage and memory.

# If you want to interact with Ollama from Python

Make sure Ollama is running on your laptop, then:

```python
import ollama

def get_response_from_model(prompt, model):
    response = ollama.generate(model=model, prompt=prompt)
    return response['response']

# Define the prompt and model
prompt = "Is Ollama the best tool I have ever encountered?"
model = "gemma2:2b"

# Get the response from the model
response = get_response_from_model(prompt, model)

# Print the formatted response
print(response)
```
