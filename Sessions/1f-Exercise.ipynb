```python
import requests

url = "https://raw.githubusercontent.com/LaurentVeyssier/Part-Of-Speech-tagging-NLP-task/refs/heads/main/brown-universal.txt"
response = requests.get(url)
response.raise_for_status()

word_pos_dict = {}

for line in response.text.strip().splitlines():
    line = line.strip()
    if not line or '\t' not in line:
        continue  # skip empty lines and sentence IDs
    word, pos = line.split('\t')
    word = word.lower()  # normalize to lowercase
    if word not in word_pos_dict:
        word_pos_dict[word] = pos
```

```python
# word_pos_dict now contains the mapping of words to their POS tags
# Example usage:
word_pos_dict['the']
```
