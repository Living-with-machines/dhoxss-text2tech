# Notebook example

### Counting characters speaking in Hamlet

Python notebook that:
1. Reads the file `hamlet.txt` in the `data/` folder,
2. Finds all occurrences of characters speaking,
3. Counts them, and
4. Returns them sorted in descending order.

```python
# Import regex module (for regular expressions)
import re
```

```python
# Load the `hamlet.txt` file from the `data/` folder,
# as a list of lines:
with open("data/hamlet.txt") as fr:
    booklines = fr.readlines()
```

Find all characters in the play:

```python
characters = []
for line in booklines:
    if re.match(r'^[A-Z ]+\..*', line):
        match = re.match(r'^([A-Z ]+)\..*', line).group(1)
        if not 'SCENE' in match:
            characters.append(match.strip())
characters = list(set(characters))
```

Print the list of characters:

```python
print("==============================")
print("List of characters")
print("==============================")
print(characters)
print()
```

Get character counts:

```python
dict_character_counts = dict()
for character in characters:
    for line in booklines:
        if line.startswith(character + "."):
            if character in dict_character_counts:
                dict_character_counts[character] += 1
            else:
                dict_character_counts[character] = 1
```

Sort characters by decreasing amount of speaking:

```python
sorted_characters = sorted(dict_character_counts.items(),key=lambda x:x[1], reverse=True)
```

Print character counts:

```python
print("==============================")
print("Character counts")
print("==============================")
for ch in sorted_characters:
    print(ch[0], ":::", ch[1])
print()
```
