## What are regular expressions

Regular expressions (regex) are **patterns that try to match specific sequences of characters** (i.e. a string). They allow us to extract information from text in a more intelligent way than by just performing exact matches.

Typical uses of regular expressions:
* Searching for patterns in text
* Counting how often a pattern occurs
* Replacing text
* Text validation: is the input well-formatted?

In the previous notebook we searched and counted with `in` and `.count()`, which only works when we know the exact string we are looking for. Regular expressions let us describe a **kind** of thing instead — a year, a page number, a capitalised name, a line of dialogue — and then find, count and replace every instance of it. This is the step that takes us from reading one text to processing a whole collection of them.

If we want to use regular expressions, we must import the ```re``` module like this:

````python
import re
````

## Some basic regex symbols:
 
* ```.``` : matches any character.
* ```*``` : previous character repeats zero or more times.
* ```+``` : previous character repeats 1 or more times.
* ```^``` : marks the start of string.
* ```$``` : marks the end of string.
* ```?``` : indicates that the previous character is optional.
* ```\d``` : matches any number.
* ```\b``` : indicates a word boundary (i.e. beginning or end of word)
* ```\.``` : matches a full stop.
* ```[A-Z]``` : matches any capital letter (A-Z) in the Latin alphabet (of the English alphabet).
* ```[a-z]``` : matches any non-capital letter (a-z) in the Latin alphabet (of the English alphabet).
* ```[abc]``` : matches any *one* of the characters listed between the brackets.
* ```{4}``` : the previous character repeats exactly four times (so `\d{4}` matches four digits in a row, e.g. a year).

## Some useful functions:
* `re.match(pattern, str)`: does a regular expression match the whole string?
* `re.findall(pattern, str)`: find all matches of a regular expression in a string.
* `re.sub(pattern, replacement, str)`: replace all instances of the found pattern in the string by the replacement.

### Matching with `re.match(pattern, str)`

````python
import re

str1 = "ACT 1, SCENE 12"
pattern = r'^ACT \d+, SCENE \d+$'

if re.match(pattern, str1):
    print(str1)
else:
    print("There's no match")
````

### Matching and capturing with ```.group()```

In the regular expression, we can specify in parentheses which is the information we would like to keep, to reuse it later.

We can then recover this information with the ```.group()``` method.

Example:

````python
import re

str1 = "ACT 1, SCENE 12"
pattern = r'^ACT (\d+), SCENE (\d+)$'

if re.match(pattern, str1):
    print("I am only interested in knowing the SCENE, which is number... " + re.match(pattern, str1).group(2))
else:
    print("There's no match")
````

Note that group `0` is the whole pattern, group `1`is the first element in parentheses, and group `2` is the second element in parentheses.

### Finding all occurrences in text with ```re.findall(pattern, str)```

````python
import re

str1 = """It was the best of times,
it was the worst of times,
it was the age of wisdom,
it was the age of foolishness,
it was the epoch of belief,
it was the epoch of incredulity."""
pattern = r'it was the (.+) of (.+),'

matches = re.findall(pattern, str1)
for match in matches:
    print(match)
````

`re.findall()` gives us back a **list** of matches — and counting the elements of a list is something we already know how to do:

````python
matches = re.findall(pattern, str1)

print(len(matches))
````

Only four? The text contains six clauses of this shape.

The first one begins with a capital `It`, and the last one ends with a full stop rather than a comma, so neither of them matches our pattern. This is the same lesson as in the previous notebook: **a count is only as good as the pattern we count with**.

Let's fix the pattern. `[Ii]` matches either an upper- or a lower-case _i_, and `[,.]` matches either a comma or a full stop:

````python
pattern = r'[Ii]t was the (.+) of (.+)[,.]'

matches = re.findall(pattern, str1)

print(len(matches))
for match in matches:
    print(match)
````

### Replacing with ```re.sub(pattern, replacement, str)```

````python
import re

str1 = """It was the best of times,
it was the worst of times,
it was the age of wisdom,
it was the age of foolishness,
it was the epoch of belief,
it was the epoch of incredulity."""
pattern = r'\bwas\b'

print(re.sub(pattern, "is not", str1))
````

# Solutions

✏️ **Exercise 1:**

````python
import re

records = """Alice's Adventures in Wonderland, London, 1865
Through the Looking-Glass, London, 1871
A Tale of Two Cities, London, 1859
Mrs Dalloway, London, 1925
The Waste Land, New York, 1922"""


# 1. Use `re.findall()` to extract all the years, and print how many you found.
years = re.findall(r'\d{4}', records)
print(years)
print(len(years))


# 2. Print how many of the records were published in London.
print(len(re.findall(r'London', records)))


# 3. Using a for-loop, count how many of the books were published before 1900.
counter = 0
for year in years:
    if int(year) < 1900:
        counter = counter + 1
print(counter)
````
