# Introduction to spaCy

spaCy is an open-source library for text preprocessing.

It features the most common components of the preprocessing pipeline:

![](images/nlp_pipeline.png)

spaCy takes a string as input, i.e. a sequence of characters, and transforms this string into a sequence of more meaningful units for analyses. Which transformations we want to perform, again, will depend on the nature of our dataset, and the type of analysis we want to perform.

Image source: https://spacy.io/usage/linguistic-features

spaCy is a python library and it has to be imported, to be able to use it:

```python
import spacy
```

### Get started with spaCy 

spaCy is built on **statistical models** that have learned to detect linguistic features from data.

spaCy provides easy-to-use pipelines for a variety of languages and tasks. For each language we'd need a different model. You can apply a model trained on English to Italian (feel free to try it out in the exercise below), but as you could guess this would not work very well!

![](images/spacy_langs.png)

Source: https://spacy.io/usage/models


We will create a pipeline using the `en_core_web_sm` model, which is trained on English data: https://spacy.io/models/en#en_core_web_sm

Often, in order to start using spacy, you'd need to download a pipeline. In our scenario, Colab has already downloaded the model, but let's do it anyway!

```python
spacy.cli.download("en_core_web_sm")
```

... And we need to import the pipeline:

```python
# Load English language pipeline and store in variable `nlp`:
nlp = spacy.load("en_core_web_sm")
```

The `nlp` pipeline takes as input a text and runs it through the pipeline.

![](images/nlp_pipeline.png)

We store the output in a variable that we call `output`:

```python
example = "This is a great week. Is it not?"
output = nlp(example)
```

If you print the content of `output`, on the surface it looks like a string. It actually looks like nothing has happened:

```python
example = "This is a great week. Is it not?"
output = nlp(example)
print(output)
```

But if we print the data type of `output`, we see it's not a string, it's a spacy object called `Doc`:

```python
# Print the type of `output`:
print(type(output))
```

Also, you'd notice that the pipeline has split the string into tokens...

```python
output[1]
```

...and tokens have certain attributes, corresponding the steps of the processing pipeline!

```python
help(output[1])
```

We can access the lemmatized form of the second token.

```python
output[1].lemma_
```

To Summarize: a [Doc](https://spacy.io/api/doc) contains the linguistic annotations as a sequences of linguistic units. The variable `output` now contains the linguistic processing of your sentence!

So all our efforts will now be put into retrieving the linguistic information from the output ([full documentation here](https://spacy.io/usage)).

## Iterating over spaCy output

How to apply spaCy for processing complete documents? For this we need the `for` loop, again...

First, to recap, that's how to process a text using spaCy:

```python
# Load English language pipeline and store in variable `nlp`. You may need to download it first,
# and you only need to do this step once!
nlp = spacy.load("en_core_web_sm")

# The text to process:
example = "This is a great week. Is it not?"

# Process the example text using the pipeline stored in `nlp`, and
# store the output in a variable called `output`. This line does most
# of the work!!!
output = nlp(example)
```

The spaCy pipeline returns a list of tokens. We can iterate over the elements in `doc` and retrieve the linguistic information.

In more technical terms, a [`Doc`](https://spacy.io/api/doc) object is a sequence of [`Token`](https://spacy.io/api/token) objects (i.e.~words). Iterating over the elements in a `Doc` object means iterating over its tokens. More interestingly, you will be able to access the token attributes, listed in https://spacy.io/api/token#attributes, using "dot notation". For example:

```python
example = "This is a great week. Is it not?"
output = nlp(example)

# Iterating over the elements in `output`, using a for-loop:
for element in output:
    # According to the documentation, .text provides the "verbatim text content" of a token. 
    print(element.text)
```

☝️It is common to use a list comprehension instead of a for-loop:

```python
example = "This is a great week. Is it not?"
output = nlp(example)

# Iterating over the elements in `output`, using a list comprehension:
print([element.text for element in output])
```
